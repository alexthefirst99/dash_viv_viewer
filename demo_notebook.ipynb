# dash_viv_viewer — Jupyter Demo

**Requirements:** The Flask backend must be running (`python sites/avivator/backend/app.py`) so images are served at `http://localhost:5001`.

1. Start the backend if not running
2. Run all cells below

```python
import dash
from dash import html, Input, Output, callback
import json
from dash_viv_viewer import VivViewer

app = dash.Dash(__name__)

IMAGE_URL = "http://127.0.0.1:5001/images/06311588-ed90-4307-b973-541198697f16.ome.tif"  # ← change to your image

app.layout = html.Div([
    html.H3('Viv Viewer with ROI Drawing', style={'fontFamily': 'sans-serif'}),
    VivViewer(
        id='viewer',
        image_url=IMAGE_URL,
        height=600,
    ),
    html.H4('ROIs (image pixel coordinates):', style={'fontFamily': 'sans-serif', 'marginTop': 12}),
    html.Pre(id='roi-output', style={'background': '#1e1e1e', 'color': '#d4d4d4', 'padding': 12, 'borderRadius': 4})
])

@callback(Output('roi-output', 'children'), Input('viewer', 'rois'))
def show_rois(rois):
    if not rois:
        return 'No ROIs drawn yet.'
    return json.dumps(rois, indent=2)

app.run(jupyter_mode='inline', port=8765)
```
